````python
import pyspark
import random
from pyspark.sql import SparkSession

# create a SparkContext to run the application
spark = SparkSession \
          .builder \
          .appName("Pi") \
          .getOrCreate()
sc = spark.sparkContext

num_samples = 100000000

# function to check if a point is inside the unit circle
def inside(p):     
  x, y = random.random(), random.random()
  return x*x + y*y < 1

# create an RDD of random points and count how many are inside the unit circle
count = sc.parallelize(range(0, num_samples)).filter(inside).count()

# calculate the value of pi using the ratio of points inside the circle to total points
pi = 4 * count / num_samples
print(pi)

# stop the SparkContext to free up resources
sc.stop()
````

## Spark SQL
To use spark SQL
- create a temporary table

```python
df.createOrReplaceTempView("tableA")

```
- run sql using the temporary table
```sql
spark.sql("SELECT count(*) from tableA").show()
```


````python
# from pyspark.sql import SparkSession

spark = SparkSession \
          .builder \
          .appName("TaxiRidesCount") \
          .getOrCreate()
sc = spark.sparkContext

input_file = "../data/yellow_tripdata_2025-01_sample.parquet"
taxi_data_df = spark.read.parquet(input_file)
````

````python
taxi_data_df.createOrReplaceTempView("yellow_taxi_data")

vendor1_count = spark.sql("SELECT COUNT(*) FROM yellow_taxi_data WHERE VendorID = 1").collect()[0][0]
vendor2_count = spark.sql("SELECT COUNT(*) FROM yellow_taxi_data WHERE VendorID = 2").collect()[0][0]
print(f"VendorID 1 count: {vendor1_count}")
print(f"VendorID 2 count: {vendor2_count}")
````

````python
df_by_vendor = spark.sql("SELECT VendorID, COUNT(*) as count FROM yellow_taxi_data GROUP BY VendorID")
df_by_vendor.show()
````
